```python
##Object oriented programming in Python 
##OOPs is a programming paradigm that uses "objects" to design software.
#An object is an instance of a class, which is a blueprint for creating objects.
```

```python
##A class is a user-defined data type that encapsulates data and functions that operate on that data.
##Classes provide a means of bundling data and functionality together.

class car:
    pass

audi = car()  ##Creating an object of class car
print(type(audi)) ##To check the type of object
print(audi) ##To check the memory location of object

bmw = car()  ##Creating another object of class car
print(type(bmw)) ##To check the type of object
print(bmw) ##To check the memory location of object
```

```python
##__init__ method is a special method that is called when an object is created. It is used to initialize the attributes of the object.
##Instance variables are variables that are defined inside a class and are unique to each object of the class.
class Dog:
    #Constructor
    def __init__(self, name, age): ##__init__ method is a constructor ##self is a reference to the current instance of the class
        self.name = name  ##Instance variable
        self.age = age    ##Instance variable

##Creating objects of class Dog
dog1 = Dog("Buddy", 3)
dog2 = Dog("Max", 5)
print(dog1.name) ##Accessing the attributes of the object dog1
print(dog1.age)
print(dog2.name) ##Accessing the attributes of the object dog2
print(dog2.age)
print(dog1) ##To check the memory location of object dog1
print(dog2) ##To check the memory location of object dog2
```

```python
##Defining methods in a class
class Dog:
    #Constructor
    def __init__(self, name, age): ##__init__ method is a constructor ##self is a reference to the current instance of the class
        self.name = name  ##Instance variable
        self.age = age    ##Instance variable

    def bark(self):  ##Method of the class Dog
        print(f"{self.name} is barking and says Woof Woof")

dog1 = Dog("Buddy", 3)
dog1.bark() ##Calling the method bark of the object dog1
dog2 = Dog("Max", 5)
dog2.bark() ##Calling the method bark of the object dog2
```

```python
#Modeling a bank account using OOPs
class BankAccount:
    ##Self is a reference to the current instance of the class and it is a key word in Python
    def __init__(self, account_number, account_holder, balance=0): ##Constructor with default value for balance
        self.account_number = account_number ##Instance variable for account number
        self.account_holder = account_holder ##Instance variable for account holder
        self.balance = balance

    def deposit(self, amount): ##amount is a instance variable for deposit which is a instance method
        if amount > 0:
            self.balance += amount
            print(f"Deposited {amount}. New balance is {self.balance}")
        else:
            print("Deposit amount must be positive")

    def withdraw(self, amount): ##amount is a instance variable for withdraw which is a instance method
        if 0 < amount <= self.balance:
            self.balance -= amount
            print(f"Withdrew {amount}. New balance is {self.balance}")
        else:
            print("Insufficient funds or invalid withdrawal amount")

    def get_balance(self): ##Method to get the current balance
        return self.balance
    
#Creating an object of BankAccount
account1 = BankAccount("123456789", "John Doe", 1000)
print(account1.get_balance()) ##Checking the balance
```

```python
##Calling the instance methods
account1.deposit(500) ##Depositing money
```

```python
account1.withdraw(200) ##Withdrawing money
print(account1.get_balance()) ##Checking the balance
```
