```python
#Method Overriding allows a child class to provide a specific implementation of a method that is already defined in its parent class.

class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return "Animal sound"
    
class Dog(Animal):
    def speak(self):  # Overriding the speak method of parent class
        return "Woof!"
class Cat(Animal):
    def speak(self):  # Overriding the speak method of parent class
        return "Meow!"
    
dog = Dog("Buddy")
cat = Cat("Whiskers")
print(dog.name + ": " + dog.speak())  # Output: Buddy: Woof!
print(cat.name + ": " + cat.speak())  # Output: Whiskers: Meow!
```

```python
#Function to demonstrate Polymorphism
def animal_sound(animal):
    print(animal.speak())
animal_sound(dog)  # Output: Woof!
animal_sound(cat)  # Output: Meow!
```

```python
##Polymorphism with functions and Methods

#Base Class
class Shape : 
    def area(self):
        return "The area of shape"
    
#Derived Class 1
class Square(Shape):
    def __init__(self, side):
        self.side = side
    def area(self):  # Overriding the area method of parent class
        return self.side * self.side
    
#Derived Class 2
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):  # Overriding the area method of parent class
        return 3.14 * self.radius * self.radius
    
##Function to demonstrate Polymorphism
def calculate_area(shape):
    print(shape.area())

square = Square(4)
circle = Circle(3)
calculate_area(square)  # Output: 16
calculate_area(circle)  # Output: 28.26
```

```python
##Polymorphism with Abstract Classes and Interfaces
"""Abstract Base Classes(ABCs) in Python are a way to define interfaces when other techniques like duck typing or hasattr() checks are not sufficient."""

from abc import ABC, abstractmethod
class Vehicle(ABC):
    @abstractmethod
    def drive(self): ##Abstract method should be empty  
        pass

class Car(Vehicle):
    def drive(self):
        return "The car is driving"
class Bike(Vehicle):
    def drive(self):
        return "The bike is driving"
    
#Function to demonstrate Polymorphism
def vehicle_drive(vehicle):
    print(vehicle.drive())  
    
#Create objects of the derived classes
car = Car()
bike = Bike()
vehicle_drive(car)  # Output: The car is driving
vehicle_drive(bike)  # Output: The bike is driving
```
