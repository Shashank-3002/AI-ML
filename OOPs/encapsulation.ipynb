```python
##Encapsulation
"""Encapsulation is one of the fundamental concepts in OOP. It describes the idea of bundling data and methods that work on that data within one unit, e.g., a class in Python. 
This concept is used to restrict access to methods and variables to prevent data from being modified by accident and to ensure that the internal representation of an object is hidden from the outside. 
This is typically done by making some attributes or methods private, meaning they cannot be accessed directly from outside the class."""


##Encapsulation with Getter and Setter methods
##Access Modifiers: Public, Protected, Private

#Example for Public Access Modifier
class Person:
    def __init__(self, name, age):
        self.name = name  ##Public attribute
        self.age = age    ##Public attribute

    def display(self):  ##Public method
        print(f"Name: {self.name}, Age: {self.age}")

person1 = Person("Alice", 30)
print(person1.name)  ##Accessing public attribute
print(person1.age)   ##Accessing public attribute
display(person1)  ##Accessing public method
```

```python
dir(person1) ##To check the attributes and methods of the class Person
```

```python
##Example for Private Access Modifier
class Employee:
    def __init__(self, name, salary):
        self.name = name          ##Public attribute
        self.__salary = salary    ##Private attribute

    def display(self):  ##Public method
        print(f"Name: {self.name}, Salary: {self.__salary}")

    def get_salary(self):  ##Getter method to access private attribute
        return self.__salary

    def set_salary(self, salary):  ##Setter method to modify private attribute
        if salary > 0:
            self.__salary = salary
        else:
            print("Invalid salary amount")

emp1 = Employee("Bob", 50000)
print(emp1.name)  ##Accessing public attribute
print(emp1.get_salary())  ##Accessing private attribute via getter method
emp1.set_salary(60000)  ##Modifying private attribute via setter method
print(emp1.get_salary())  ##Accessing modified private attribute via getter method
emp1.set_salary(-1000)  ##Trying to set an invalid salary amount
```

```python
dir(emp1) ##To check the attributes and methods of the class Employee
```

```python
##Protected Access Modifier

class BankAccount:
    def __init__(self, account_number, balance):
        self.account_number = account_number  ##Public attribute
        self._balance = balance                ##Protected attribute

    def deposit(self, amount):  ##Public method
        if amount > 0:
            self._balance += amount
            print(f"Deposited: {amount}")
        else:
            print("Invalid deposit amount")

    def withdraw(self, amount):  ##Public method
        if 0 < amount <= self._balance:
            self._balance -= amount
            print(f"Withdrew: {amount}")
        else:
            print("Invalid withdrawal amount or insufficient funds")

    def get_balance(self):  ##Public method to access protected attribute
        return self._balance
    


##Derived class to access protected attribute
class SavingsAccount(BankAccount):
    def __init__(self, account_number, balance, interest_rate):
        super().__init__(account_number, balance)
        self.interest_rate = interest_rate  ##Public attribute

    def apply_interest(self):  ##Public method
        interest = self._balance * self.interest_rate / 100
        self._balance += interest
        print(f"Applied interest: {interest}")




##Creating object of class BankAccount
account1 = BankAccount("123456789", 1000)
print(account1.account_number) ##Accessing public attribute
print(account1.get_balance()) ##Accessing protected attribute via public method
print(account1._balance) ##Accessing protected attribute directly (not recommended)


##Creating object of derived class SavingsAccount
savings_account = SavingsAccount("987654321", 2000, 5)
print(savings_account.account_number) ##Accessing public attribute
print(savings_account.get_balance()) ##Accessing protected attribute via public method
print(savings_account._balance) ##Accessing protected attribute directly (not recommended)
savings_account.apply_interest() ##Applying interest to the savings account
print(savings_account.get_balance()) ##Checking the balance after applying interest
```
