```python
##Abstraction in Python
"""Abstraction is the concept of hiding the complex implementation details and showing only the essential features of the object. 
It helps in reducing complexity and increases efficiency. 
In Python, abstraction can be achieved using abstract classes and interfaces."""

from abc import ABC, abstractmethod

#Abstract Base class
class Vehicle(ABC):
    def drive(self): ##Public method 
        print("The vehicle is driving.")  #Common functionality for all vehicles

    @abstractmethod  ##Abstract method which the derived class must implement with its own functionality
    def fuel_type(self): ##Abstract method
        pass

#Derived class Car with specific functionality for the abstract method
class Car(Vehicle):
    def fuel_type(self):
        # return super().fuel_type()  # Implementing the abstract method
        print("The car runs on petrol or diesel.")  #Specific functionality for Car


#Derived class Bike with specific functionality for the abstract method
class Bike(Vehicle):
    def fuel_type(self):
        # return super().fuel_type()  # Implementing the abstract method
        print("The bike runs on petrol.")  #Specific functionality for Bike

#Creating objects of the derived classes
car = Car()
bike = Bike()
car.drive()  # Output: The vehicle is driving.
car.fuel_type()  # Output: The car runs on petrol or diesel.
bike.drive()  # Output: The vehicle is driving.
bike.fuel_type()  # Output: The bike runs on petrol.
```

```python
##Magic Methods in Python
"""Magic methods in Python are special methods that have double underscores at the beginning and end of their names."""
"""These methods are also known as dunder methods (double underscore methods)."""

##Examples of magic methods include __init__, __str__, __repr__, __add__, __len__, etc.

##Example of a class with magic methods
class Person:
    def __init__(self, name, age):  ##Constructor method to initialize the object
        self.name = name
        self.age = age

    def __str__(self):  ##String representation of the object
        return f"Person(Name: {self.name}, Age: {self.age})"

    def __len__(self):  ##Method to return the length of the name
        return len(self.name)
    
#Creating an object of the Person class
person = Person("Alice", 30)
print(person)  # Output: Person(Name: Alice, Age: 30)
print(len(person))  # Output: 5 (length of the name "Alice")
```

```python
##Operator Overloading in Python
"""Operator overloading in Python allows you to define custom behavior for operators (like +, -, *, etc.) for your own classes."""

##Example of operator overloading
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):  ##Overloading the + operator
        return Vector(self.x + other.x, self.y + other.y)
    
    def __sub__(self, other):  ##Overloading the - operator
        return Vector(self.x - other.x, self.y - other.y)
    
    def __mul__(self, scalar):  ##Overloading the * operator for scalar multiplication
        return Vector(self.x * scalar, self.y * scalar)
    
    def __eq__(self, value):
        return self.x == value.x and self.y == value.y  ##Overloading the == operator

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"  ##Official string representation of the object
    
    def __str__(self):  ##String representation of the object
        return f"Vector({self.x}, {self.y})"   
#Creating objects of the Vector class
v1 = Vector(2, 3)
v2 = Vector(4, 5)
v3 = v1 + v2  # Using the overloaded + operator
print(v3)  # Output: Vector(6, 8)
print(v1 - v2)  # Output: Vector(-2, -2) (Using the overloaded - operato)
print(v1 * 3)  # Output: Vector(6, 9) (Using the overloaded * operator for scalar multiplication)
print(v1 == v2)  # Output: False (Using the overloaded == operator)
```

```python
##Overloading operators for complex numbers

class ComplexNumber:
    def __init__(self, real, imag):
        self.real = real
        self.imag = imag

    def __add__(self, other):  ##Overloading the + operator
        return ComplexNumber(self.real + other.real, self.imag + other.imag)
    
    def __sub__(self, other):  ##Overloading the - operator
        return ComplexNumber(self.real - other.real, self.imag - other.imag)
    
    def __mul__(self, other):  ##Overloading the * operator
        real_part = self.real * other.real - self.imag * other.imag
        imag_part = self.real * other.imag + self.imag * other.real
        return ComplexNumber(real_part, imag_part)
    
    def __repr__(self):
        return f"{self.real} + {self.imag}i"  ##Official string representation of the object
    
    def __str__(self):  ##String representation of the object
        return f"{self.real} + {self.imag}i"    
#Creating objects of the ComplexNumber class
c1 = ComplexNumber(2, 8)
c2 = ComplexNumber(4, 5)
print(c1 + c2)  # Output: 6 + 13i (Using the overloaded + operator)
print(c1 - c2)  # Output: -2 +3i (Using the overloaded - operator)
print(c1 * c2)  # Output: -32 + 42i (Using the overloaded * operator)
print(c1)  # Output: 2 + 8i (Using the __str__ method)
```
