```python
##Inheritance in python is defined as the capability of one class to derive or inherit the properties from another class.

#Example of a class car
class car:
    def __init__(self, brand, model, year): ##__init__ method is a constructor ##self is a reference to the current instance of the class
        self.brand = brand  ##Instance variable or attributes 
        self.model = model  ##Instance variable
        self.year = year    ##Instance variable

    def drive(self): ##Instance method
        print(f"The person will drive the {self.brand} {self.model} car of {self.year} model.")

##Creating objects of class car
car1 = car("Toyota", "Camry", 2020)
car2 = car("Honda", "Civic", 2019)
print(car1.brand) ##Accessing the attributes of the object car1
print(car1.model)
print(car2.brand) ##Accessing the attributes of the object car2
print(car2.model)

car1.drive() ##Calling the instance method
car2.drive() ##Calling the instance method
```

```python
##Single Inheritance Example
class ElectricCar(car): ##ElectricCar is a child class which inherits from car class
    def __init__(self, brand, model, year, battery_size): ##__init__ method is a constructor ##self is a reference to the current instance of the class
        super().__init__(brand, model, year) ##Calling the constructor of the parent class
        self.battery_size = battery_size ##Instance variable or attributes 

    def charge(self): ##Instance method
        print(f"The {self.brand} {self.model} car is charging with a battery size of {self.battery_size} kWh.")

    def drive(self):
        super().drive()
        print(f"The {self.brand} {self.model} car drives with the battery capacity of {self.battery_size} kW.")  # Overriding the instance method of parent class
##Creating objects of class ElectricCar
electric_car1 = ElectricCar("Tesla", "Model S", 2021, 100)
print(electric_car1.brand) ##Accessing the attributes of the object electric_car1
print(electric_car1.model)
print(electric_car1.battery_size) ##Accessing the attributes of the object electric_car1
electric_car1.drive() ##Calling the instance method of parent class
```

```python
##Example for multiple inheritance

class Animal: ##Base class or Parent class 1
    def __init__(self, name, species):
        self.name = name
        self.species = species

    def make_sound(self):
        print(f"The {self.species} named {self.name} makes a sound.")

class Pet(): ##Base class or Parent class 2
    def __init__(self, owner):
        self.owner = owner

    def play(self):
        print(f"{self.name} is playing with its owner {self.owner}.")

class Dog(Animal, Pet): ##Derived class or Child class inheriting from Animal and Pet classes
    def __init__(self, name, species, owner):
        Animal.__init__(self, name, species)
        Pet.__init__(self, owner)

    def bark(self):
        print(f"{self.name} says Woof!")

##Creating objects of class Dog
dog1 = Dog("Buddy", "Dog", "Alice")
print(dog1.name) ##Accessing the attributes of the object dog1
print(dog1.species) ##Accessing the attributes of the object dog1
print(dog1.owner) ##Accessing the attributes of the object dog1
dog1.make_sound() ##Calling the instance method of parent class Animal
dog1.play() ##Calling the instance method of parent class Pet
dog1.bark() ##Calling the instance method of class Dog
```
