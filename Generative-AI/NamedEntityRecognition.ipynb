### Named Entity Recognition

```python
sentence="The Eiffel Tower was built from 1887 to 1889 by Gustave Eiffel, whose company specialized in building metal frameworks and structures."
```

```python
import nltk
words=nltk.word_tokenize(sentence)
```

```python
tag_elements=nltk.pos_tag(words)
tag_elements
```

```python
nltk.download('maxent_ne_chunker')
```

```python
nltk.download('words')
```

```python
"""NER stands for Named Entity Recognition. 
It is a subtask of Natural Language Processing (NLP) that focuses on identifying and classifying named entities in text into predefined categories 
such as person names, organizations, locations, dates, and more. 
The goal of NER is to extract meaningful information from unstructured text by recognizing and categorizing specific entities."""
nltk.ne_chunk(tag_elements).draw()
```
