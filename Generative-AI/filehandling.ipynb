```python
##Reading a file

with open('example.txt', 'r') as file:
    content = file.read()
    print(content)
```

```python
#Read a file line by line
with open('example.txt', 'r') as file:
    for line in file:
        print(line.strip())  # strip() removes leading/trailing whitespace and newlines
```

```python
#Write to a file with 'w' mode and without overwriting existing content  with 'a' mode
with open('output.txt', 'w') as file: ##'w' mode is for writing, 'a' mode is for appending
    file.write("Hello, World!\n")
    file.write("This is a test file.\n")
```

```python
##Append to a file
with open('output.txt', 'a') as file:
    file.write("Appending a new line.\n")
    file.write("Another appended line.\n")
```

```python
##Writing a list of lines to a file
lines = ["First line\n", "Second line\n", "Third line\n"]
with open('output.txt', 'a') as file:
    file.writelines(lines)
```

```python
##Binary file handling
with open('example.bin', 'wb') as bin_file:
    bin_file.write(b'\x00\xFF\x7A\x3C')  # Writing some binary data
with open('example.bin', 'rb') as bin_file:
    binary_content = bin_file.read()
    print("Binary content:", binary_content)
```

```python
#Read the content from a source file and write it to a destination file
with open('example.txt', 'r') as src_file:
    content = src_file.read()
with open('copy_of_example.txt', 'w') as dest_file:
    dest_file.write(content)
```

```python
##Read a text file and count the number of lines, words, and characters
with open('example.txt', 'r') as file:
    lines = file.readlines()
    num_lines = len(lines)
    num_words = sum(len(line.split()) for line in lines)
    num_chars = sum(len(line) for line in lines)
    print(f"Lines: {num_lines}, Words: {num_words}, Characters: {num_chars}")
```

```python
## Create a new directory 
import os
os.makedirs('new_directory')
```

```python
## List files in a directory
files = os.listdir('.')
print("Files in current directory:", files)
```

```python
##Listing only .txt files in a directory
txt_files = [f for f in os.listdir('.') if f.endswith('.txt')]
print(".txt files in current directory:", txt_files)
```

```python
##Move the .txt files to the new directory
import shutil
for txt_file in txt_files:
    shutil.move(txt_file, os.path.join('new_directory', txt_file))
    
```

```python
##Checking a file's existence
if os.path.exists('example.txt'):
    print("example.txt exists.")
else:
    print("example.txt does not exist.")
```

```python
##Checking if a path is a file or directory
if os.path.isfile('example.txt'):
    print("example.txt is a file.")
if os.path.isdir('new_directory'):
    print("new_directory is a directory.")
    
```

```python
##Getting the absolute path of a file
abs_path = os.path.abspath('example.txt')
print("Absolute path of example.txt:", abs_path)
```
