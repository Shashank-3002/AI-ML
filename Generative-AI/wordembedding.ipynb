```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sentence = ['I love my country',
            'I am a good boy',
            'I am a good developer',
            'I am a awesome person',
            'I am a best friend']
```

```python
sentence
```

```python
##Define the vocabulary size
vocab_size = 5000
```

```python
##One hot representation
one_hot_representations = [one_hot(words, vocab_size) for words in sentence]
one_hot_representations
##This is gives the output as a list of list. Each word is represented as a number. This number is the index of the word in the vocabulary. The vocabulary is created by the one_hot function. The one_hot function takes the sentence and the vocabulary size as input and returns the one hot representation of the sentence.
```

```python
##word embedding representation
from tensorflow.keras.layers import Embedding
from tensorflow.keras.preprocessing.sequence import pad_sequences #To make all the sentences of the same length
from tensorflow.keras.models import Sequential #To create a sequential model
```

```python
import numpy as np
```

```python
sentence_length = 8
embedded_docs = pad_sequences(one_hot_representations, padding='pre', maxlen=sentence_length)
embedded_docs
```

```python
##Feature representation
dimension = 10
model = Sequential()
model.add(Embedding(vocab_size, dimension, input_length=sentence_length))
model.compile('adam', 'mse') ##To compile the model we need to specify the optimizer and the loss function. Here we are using 'adam' as the optimizer and 'mse' as the loss function. The 'adam' optimizer is a popular optimization algorithm that is used in deep learning. The 'mse' loss function is used for regression problems. It stands for mean squared error. It calculates the average of the squares of the errors between the predicted values and the actual values.
```

```python
model.summary()
```

```python
model.predict(embedded_docs)
```
