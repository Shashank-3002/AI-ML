```python
#Example 1: Handling ZeroDivisionError
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")
```

```python
#Example 2: Handling FileNotFoundError
try:
    with open('non_existent_file.txt', 'r') as file:
        content = file.read()
except FileNotFoundError:
    print("Error: The specified file was not found.")
```

```python
try:
    num = int(input("Enter a number: "))
    result = 10 / num
except ValueError:
    print("Error: Invalid input. Please enter a valid integer.")
except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")
except Exception as e: ##To handle any other unexpected exceptions
    print(f"An unexpected error occurred: {e}")
```

```python
##try, except, else block
try:
    num = int(input("Enter a number: "))
    result = 10 / num
except ValueError:
    print("Error: Invalid input. Please enter a valid integer.")
except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")
except Exception as e: ##To handle any other unexpected exceptions
    print(f"An unexpected error occurred: {e}")
else: ##executes if no exception occurs
    print(f"The result is: {result}")
```

```python
##Finally block: The finally block is used to define a block of code that will always execute, regardless of whether an exception occurred or not. It is typically used for cleanup actions, such as closing files or releasing resources.
try:
    num = int(input("Enter a number: "))
    result = 10 / num
except ValueError:
    print("Error: Invalid input. Please enter a valid integer.")
except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")
except Exception as e: ##To handle any other unexpected exceptions
    print(f"An unexpected error occurred: {e}")
else: ##executes if no exception occurs
    print(f"The result is: {result}")
finally: ##executes always
    print("Execution completed.")

##Finally block is used to ensure that certain cleanup actions are performed, regardless of whether an exception occurred or not.
#It is used to define a block of code that will always execute, making it useful for tasks like closing files or releasing resources.
```
