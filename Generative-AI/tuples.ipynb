```python
numbers = (1,2,3,4,5)
print(numbers)
print(type(numbers))
```

```python
mixed_tpl = (1,2,"Hello World",3.142, True)
mixed_tpl
```

```python
print(type(mixed_tpl))
```

```python
##Accessing the tuples
print(mixed_tpl[2])
print(mixed_tpl[:2])
print(mixed_tpl[2:])
print(mixed_tpl[:-1])
print(mixed_tpl[-3:])
print(mixed_tpl[0:2])
print(mixed_tpl[2:4])
```

```python
##Type of the elements of the tuple
print(type(mixed_tpl[2])) #Hello World
print(type(mixed_tpl[3])) #3.142
print(type(mixed_tpl[4])) #True
```

```python
##Tuple Operations

##Concatination of two tuples
# con_tuple = numbers + mixed_tpl[2:]
con_tuple = numbers + mixed_tpl
print(con_tuple)
print(type(con_tuple))
```

```python
mixed_tpl*3
```

```python
numbers*3
```

```python
##Immutable nature of tuples --> The elements in the tuples cannot be changed once assigned

##Example of list --> Mutable elements
lst = [10,20,30,40,50]
print(lst)

lst[1] = "Shashank"
print(lst)

##Changing the element in the numbers tuple

# numbers[2] = "Shashank" 
#TypeError: 'tuple' object does not support item assignment
```

```python
#Tuple methods
print(con_tuple)
print(con_tuple.count(1))
print(con_tuple.index(2))
```

```python
#Packing and unpacking the tuple

#Packing the tuples
packed_tuple = 1,"Hii",2.141
print(packed_tuple)

#Unpacking the tuples
a,b,c = packed_tuple
print(a)
print(b)
print(c)

#Unpacking the tuples with *
first,*middle,last = numbers
print(first)
print(middle)
print(last)
```

```python
##Nested tuples

nested = ((numbers),(mixed_tpl),(con_tuple),(packed_tuple))
print(nested)

print(nested[0],"\n")
print(nested[1],"\n")
print(nested[2],"\n")
print(nested[3],"\n")
print(nested[0:2],"\n")
print(nested[2:],"\n")
print(nested[0][1],"\n")
print(nested[1][2],"\n")
print(nested[2][3],"\n")
print(nested[3][1],"\n")
```
