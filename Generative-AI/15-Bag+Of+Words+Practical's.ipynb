```python
import pandas as pd

messages = pd.read_csv(
    r'C:\Users\shash\Desktop\SamsungInnovationCampus-AI-ML\SamsungInnovationCampus-AI-ML\DataSets\spam.csv',
    encoding='latin-1'
)
messages = messages[['v1','v2']]
messages.columns = ['label','message']
```

```python
messages
```

```python
## Data Cleaning And Preprocessing
import re #For regular expression operations
import nltk
nltk.download('stopwords') #For downloading the stop words from the nltk library
```

```python
from nltk.corpus import stopwords #For stop words which are commonly used words in a language that do not add much meaning to a sentence and can be removed without losing the context of the sentence
from nltk.stem.porter import PorterStemmer #For stemming which is the process of reducing a word to its root form. For example, the words "running", "runner", and "ran" can be reduced to the root word "run". This helps in reducing the dimensionality of the data and improving the performance of machine learning models.
ps=PorterStemmer()
```

```python
corpus=[] #Creating an empty list to store the cleaned reviews
for i in range(0,len(messages)):
    review=re.sub('[^a-zA-z]',' ',messages['message'][i]) #Substituting all the characters other than a-z and A-Z with space
    review=review.lower() #Converting all the characters to lower case
    review=review.split() #Splitting the review into list of words
    #Stemming and removing stop words from the review and then joining the words back to form a cleaned review and appending it to the corpus list
    review=[ps.stem(word) for word in review if not word in stopwords.words('english')]
    review=' '.join(review)
    corpus.append(review)
```

```python
corpus
```

## Create Bag Of Words

```python
## Create the Bag OF Words model
from sklearn.feature_extraction.text import CountVectorizer
## for Binary BOW enable binary=True

"""Counter Vectorizer is a class in the sklearn library that is used to convert a collection of text documents to a matrix of token counts. 
It creates a bag of words model where each row represents a document and each column represents a word in the vocabulary. 
The value in each cell represents the count of the word in the document. By setting binary=True, we can create a binary bag of words model 
where the value in each cell is 1 if the word is present in the document and 0 if it is not present. 
This can help in reducing the dimensionality of the data and improving the performance of machine learning models."""


cv=CountVectorizer(max_features=100,binary=True) #Creating a CountVectorizer object with a maximum of 100 features and binary=True to create a binary bag of words model
X=cv.fit_transform(corpus).toarray() #Fitting the CountVectorizer object to the corpus and transforming it into a sparse matrix and then converting it to an array
```

##### Ngram is a contiguous sequence of n items from a given sample of text or speech. The items can be phonemes, syllables, letters, words or base pairs according to the application. The n-grams typically are collected from a text or speech corpus. When the items are words, n-grams may also be called shingles. 
###### Unigram is a contiguous sequence of 1 item
###### Digram is a contiguous sequence of 2 items
###### Trigram is a contiguous sequence of 3 items

```python
## Create the Bag OF Words model with ngram
from sklearn.feature_extraction.text import CountVectorizer
## for Binary BOW enable binary=True
"""CounterVectorizer(ngram_range=(2,3)) will create a bag of words model with ngrams of size 2 and 3. 
For example, if the review is "I love this movie", then the bigrams will be "I love", "love this", "this movie" and 
the trigrams will be "I love this", "love this movie".
This will help in capturing the context of the words in the review and improving the performance of machine learning models."""

cv=CountVectorizer(max_features=500,binary=True,ngram_range=(2,3)) #Creating a CountVectorizer object with a maximum of 100 features, binary=True to create a binary bag of words model and ngram_range=(2,3) to create a bag of words model with ngrams of size 2 and 3
X=cv.fit_transform(corpus).toarray()
```

```python
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet') #For downloading the wordnet from the nltk library
wordlematizer=WordNetLemmatizer() #Creating a WordNetLemmatizer object
```

```python
corpus=[] #Creating an empty list to store the cleaned reviews
for i in range(0,len(messages)):
    review=re.sub('[^a-zA-z]',' ',messages['message'][i]) #Substituting all the characters other than a-z and A-Z with space
    review=review.lower() #Converting all the characters to lower case
    review=review.split() #Splitting the review into list of words
    #Lemmatizing and removing stop words from the review and then joining the words back to form a cleaned review and appending it to the corpus list
    review=[wordlematizer.lemmatize(word) for word in review if not word in stopwords.words('english')]
    review=' '.join(review)
    corpus.append(review)
```
