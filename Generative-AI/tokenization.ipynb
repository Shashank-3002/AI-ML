```python
corpus="""Hello Welcome,to Krish Naik's NLP Tutorials.
Please do watch the entire course! to become expert in NLP.
"""
```

```python
print(corpus)
```

```python
##  Tokenization
## Sentence-->paragraphs
from nltk.tokenize import sent_tokenize
```

```python
documents=sent_tokenize(corpus)
```

```python
#Type of documents is list because it contains multiple sentences. Each sentence is an element of the list.
type(documents)
```

```python
for sentence in documents:
    print(sentence)
```

```python
## Tokenization 
## Paragraph-->words
## sentence--->words
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
for sentence in documents:
    print(word_tokenize(sentence))
#Prints the list of words in each sentence. Each sentence is an element of the list. Each word is an element of the list.   
```

```python
#Tokenization using wordpunct_tokenize 
from nltk.tokenize import wordpunct_tokenize
```

```python
wordpunct_tokenize(corpus) #Splits the text into words and punctuation. Each word and punctuation is an element of the list.
```

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
tokenizer=TreebankWordTokenizer()
```

```python
tokenizer.tokenize(corpus)
```
