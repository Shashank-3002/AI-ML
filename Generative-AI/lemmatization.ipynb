## Wordnet Lemmatizer

##### Lemmatization technique is like stemming. The output we will get after lemmatization is called 'lemma', which is a root word rather than root stem, the output of stemming.

```python
"""After lemmatization, we will be getting a valid word that means the same thing.
NLTK provides WordNetLemmatizer class which is a thin wrapper around the wordnet corpus. This class uses morphy() function to the WordNet CorpusReader class to find a lemma. Let us understand it with an example"""
```

```python
import nltk
```

```python
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer
lemma = WordNetLemmatizer() ##Lemma is an instance of WordNetLemmatizer class

#Different functions of lemmatization
# 1. lemmatize(word, pos='n') --> returns the lemma of the word
# 2. lemmatize(word) --> returns the lemma of the word with default post tag(pos) as noun
##Functionality of lemmatization
print(lemma.lemmatize("running", pos='v'))  #pos parameter specifies the part of speech; 'v' for verb
print(lemma.lemmatize("better", pos='a'))   # 'a' for adjective
print(lemma.lemmatize("cats"))               #default pos is noun
print(lemma.lemmatize("going", pos = 'v'))
```

```python
words = ["running", "runner", "ran", "easily", "fairly", "fishing", "fished", "fisher", "cats", "caring", "careful", "carefully"]

for w in words:
    print(w+"---->"+lemma.lemmatize(w)) #pos is set to default as noun
```

```python
for w in words:
    print(w+"---->"+lemma.lemmatize(w,pos='a')) #pos=adjective
```

```python
for w in words:
    print(w+"---->"+lemma.lemmatize(w,pos='v')) #pos=verb
```

```python
for w in words:
    print(w+"---->"+lemma.lemmatize(w,pos='r')) #adverb
```
