    Moduls

2. Comparision operators

```python
import numpy as np
```

```python
#Arithmetic Operations
a = 10
b =15
#ADD
add = a+b
print(add)
#SUBTRACTION
sub = a-b
print(sub)
#MULTIPLICATION
mul = a*b
print(mul)
#DIVISION   
div = a/b
print(div)
#MODULUS
mod = a%b
print(mod)
#FLOOR DIVISION
Fdiv = b//a
print(Fdiv)
#EXPONENTIATION
exp_a = a**3
exp_b = b**3
print(exp_a)
print(exp_b)
```

```python
#Comparision Operators --> Compares the two varieable's literals and returns the boolean value i.e., True or False

#Equals to (==)
print(a == b)

#Not equals to(!=)
print(a!=b)

#Greater than (>)
print(a>b)

#lesser than (<)
print(a<b)

#Greater than or equal to(>=)
print(a>=b)

#Lesser than or equal to (<=)
print(a<=b)
```

###Logical Operator

```python
#AND OPERATOR
x = True
y = True
z = False

and_result_1 = x and y
print(and_result_1)
and_result_2 = x and z
print(and_result_2)
and_result_3 = y and z
print(and_result_3)
```

```python
#OR     operator
or_result_1 = x or y
print(or_result_1)
or_result_2 = x or z
print(or_result_1)
or_result_3 = z or y
print(or_result_1)
```

```python
#NOT OPERATOR
not_1 = not(x)
print(not_1)
not_2 = not(y)
print(not_2)
not_3 = not(z)
print(not_3)
```
