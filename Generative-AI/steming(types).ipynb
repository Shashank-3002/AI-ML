#### Stemming is the process of reducing a word to its word stem that affixes to suffixes and prefixes or to the roots of words known as a lemma. Stemming is important in natural language understanding (NLU) and natural language processing (NLP).
#### Definition: Stemming reduces words to their core stem or lemma. 
#### Function: The base form is where suffixes and prefixes are attached.
#### Application: It is a key process in NLU and NLP.
#### Purpose: Helps computers understand and process human language more effectively.

```python
## Classification problem
##comments of a product are classified as positive or negative
## Reviews with rating >= 4 are positive, others are negative
```

```python
from nltk.stem import PorterStemmer

words = ["running", "runner", "ran", "easily", "fairly", "fishing", "fished", "fisher", "cats", "caring", "careful", "carefully"]
ps = PorterStemmer()

for w in words:
    print(f"{w} --> {ps.stem(w)}")
```

```python
ps.stem("congratulations")
##Disadvantages:
## 1. Sometimes the root word is not a valid word
## 2. Different words can have same root word
## E.g., "universe" and "university" both get stemmed to "univers"
## 3. It may remove important suffixes that change the meaning of the word
## E.g., "better" gets stemmed to "better" losing the comparative meaning
```

```python
##RegexpStemmer---> NLTK has RegexpStemmer which stems words based on a regular expression pattern and we can easily define our own stemming rules and implement regular expressions for stemming.
from nltk.stem import RegexpStemmer
rs = RegexpStemmer('ing$|s$|e$|able$|ily$', min=6)
rs.stem("running") #removes 'ing' suffix
```

```python
rs.stem("easily") #removes 'ly' suffix
```

```python
rs.stem("fairly") #removes 'ly' suffix
```

### Snowball Stemmer

```python
##SnowballStemmer ---> It is an improved version of the Porter Stemmer and supports multiple languages. It is more aggressive in stemming compared to the Porter Stemmer.
from nltk.stem import SnowballStemmer
ss = SnowballStemmer("english")
for w in words:
    print(f"{w} --> {ss.stem(w)}")
```
