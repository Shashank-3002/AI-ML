```python
##Data Collection
import nltk
nltk.download('gutenberg')
from nltk.corpus import gutenberg
import pandas as pd

# Load the text data from the Gutenberg corpus
text_data = gutenberg.raw('shakespeare-hamlet.txt')

##Save to file
with open('hamlet.txt', 'w') as file:
    file.write(text_data)
```

```python
##Data Preprocessing
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer ##For converting text to sequences
from tensorflow.keras.preprocessing.sequence import pad_sequences #For padding sequences to a fixed length
from sklearn.model_selection import train_test_split

#load the text data
with open('hamlet.txt', 'r') as file:
    text_data = file.read().lower() # Convert to lowercase

# Tokenize the text
tokenizer = Tokenizer(num_words=10000, oov_token="<OOV>")
tokenizer.fit_on_texts([text_data]) # Create sequences of words
total_words = len(tokenizer.word_index) + 1 # Total number of unique words + 1 for padding
```

```python
total_words
```

```python
tokenizer.word_index.items() #View the word index mapping (word to integer)
```

```python
##Create input sequences
input_sequences = []
for line in text_data.split('.'):
    token_list = tokenizer.texts_to_sequences([line])[0] # Convert line to sequence of integers
    for i in range(1, len(token_list)):
        n_gram_sequence = token_list[:i+1] # Create n-gram sequences
        input_sequences.append(n_gram_sequence)
```

```python
input_sequences #View the generated input sequences which display the n-gram sequences of integers representing the words in the text. Each sequence starts with the first word and adds one more word at a time until the end of the line is reached.
```

```python
##Padding sequences to a fixed length
max_sequence_len = max([len(x) for x in input_sequences]) # Find the maximum sequence length
max_sequence_len
```

```python
input_sequences = np.array(pad_sequences(input_sequences, maxlen=max_sequence_len, padding='pre')) # Pad sequences to the maximum length
input_sequences #View the padded input sequences which are now all of the same length, with shorter sequences padded with zeros at the beginning.
```

```python
##Create Predictors and Labels
import tensorflow as tf
#x = input_sequences[:,:-1] # All elements except the last one in each sequence (predictors)
#y = input_sequences[:,-1] # The last element in each sequence (label)
x, y = input_sequences[:,:-1], input_sequences[:,-1] # Split sequences into predictors (x) and labels (y)
```

```python
y
```

```python
y = tf.keras.utils.to_categorical(y, num_classes=total_words)
```

```python
input_sequences = np.array(input_sequences)

x = input_sequences[:, :-1]
y = input_sequences[:, -1]   # keep as integers
```

```python
#Split the data into training and testing sets
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
print(x_train.shape)
print(y_train.shape)
```

```python
from tensorflow.keras.callbacks import EarlyStopping
early_stopping = EarlyStopping(monitor='val_loss', patience=3, restore_best_weights=True)
```

```python
##Train the LSTM model
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout, GRU
from tensorflow.keras.optimizers import Adam

optimizer = Adam(learning_rate=0.001)

model = Sequential()
model.add(Embedding(total_words, 64, input_length=max_sequence_len-1)) # Embedding layer to convert integer sequences to dense vectors
model.add(LSTM(128, return_sequences = True))
model.add(Dropout(0.2))
model.add(LSTM(128))
model.add(Dense(total_words, activation="softmax"))

#Compile the model
model.compile(loss = "sparse_categorical_crossentropy", optimizer = optimizer, metrics = ['accuracy'])
model.summary()
```

```python
##Train the model
history = model.fit(x_train, y_train, epochs=25, validation_data=(x_test, y_test), verbose=1, callbacks=[early_stopping])
```
