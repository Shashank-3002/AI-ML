##### One hot encoding is a technique used to convert categorical data into a binary format that can be easily understood by machine learning algorithms. It creates a binary column for each category and assigns a value of 1 to the column corresponding to the category and 0 to all other columns. This allows the model to learn from the categorical data without assuming any ordinal relationship between the categories.

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

# 1. Create a sample dataset
data = {'Device': ['Mobile', 'Desktop', 'Tablet', 'Mobile']}
df = pd.DataFrame(data)

# 2. Initialize the OneHotEncoder
# 'sparse_output=False' makes it easy to read as an array
encoder = OneHotEncoder(sparse_output=False)

# 3. Fit and transform the data
encoded_data = encoder.fit_transform(df[['Device']])

# 4. Create a readable DataFrame with the new column names
encoded_df = pd.DataFrame(encoded_data, columns=encoder.get_feature_names_out(['Device']))

print("Original Data:\n", df)
print("\nEncoded Data:\n", encoded_df)
```
