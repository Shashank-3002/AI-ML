```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
A = pd.DataFrame({'Name' : ['Shashank', 'Rohit', 'Ankit', 'Sagar','Vivek'],
                  'Age' : [21, 22, 21, 23, 22],
                  'Gender' : ['Male', 'Male', 'Male', 'Male', 'Male']})
```

```python
A
```

```python
B = pd.DataFrame({'Name': ['Shashank', 'Ankita','Rohit'],
                  'Position': ['Data Scientist', 'Developer', 'Manager'],
                  'Salary': [100000, 80000, 120000]})
```

```python
B
```

```python
#Simple Merging DataFrames on Name column for common values 
pd.merge(A, B, on='Name')
```

```python
pd.merge(A, B, how='left', on='Name')
```

```python
pd.merge(A, B, how='right', on='Name')
```

```python
pd.merge(A,B,on='Name', how='outer')
```

```python
pd.merge(A,B, on = 'Name', how='inner')
```
