```python
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df = pd.read_csv('iris.csv', header='infer')
```

```python
df.shape
```

```python
df.head()
```

```python
df.drop(columns=['Id'], inplace=True)
```

```python
df.loc[:, 'PetalLengthCm'].plot.hist(bins = 20, color='c', edgecolor='k', alpha=0.7)
plt.title('Petal Length Distribution')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Frequency')
plt.show()
```

```python
df.loc[:, ['PetalLengthCm', 'PetalWidthCm']].plot.hist(bins = 50, color=['red','green'], alpha=0.5)
plt.show()
```

```python
frequencies = df.Species.value_counts()
my_counts = list(frequencies.values)
my_labels = list(frequencies.index)
df1 = pd.DataFrame({'counts':my_counts}, index=my_labels)
df1
```

```python
df1.plot.bar(rot=0, color=['c', 'm', 'y'], alpha=0.7)
plt.show()
```

```python
df.plot.scatter(x='PetalLengthCm', y='PetalWidthCm', color='c', alpha=0.7, marker='o', s=50)
plt.show()
```

```python
my_cols_dict = {'Iris-setosa':'g', 'Iris-versicolor':'r', 'Iris-virginica':'y'}
my_cols = df['Species'].apply(lambda x: my_cols_dict[x])
pd.plotting.scatter_matrix(df ,c=my_cols, alpha=0.7, marker='o', s=50, figsize=(10,10))
plt.show()
```
