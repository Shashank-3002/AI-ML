```python
student1 = pd.Series({'Korean': 90, 'Maths': 80, 'Science': 85}) 
```

```python
student1
```

```python
Percentage = (student1 / 100) * 100 #Percentage of each subject marks
```

```python
Percentage
```

```python
Total_percentage = (student1.sum()/ 300)*100 #Total Percentage
Total_percentage
```

```python
add = student1 + 5
add
```

```python
sub = student1 - 10
sub
```

```python
div = student1 / 10
div
```

```python
#Creating new data set from the operations on the existing data set
df_1 = pd.DataFrame(data = [add, sub, div], index = ['Addition', 'Subtraction', 'Division'])
df_1
```

```python
student2 = pd.Series({'Korean': 80, 'Maths': 75, 'Pysics': 90})
student2
```

```python
mul = student1 * student2
mul
```

```python
df_2 = pd.DataFrame(data = [add, sub, div, mul], index = ['Addition', 'Subtraction', 'Division', 'Multiplication'])
df_2
```

```python
#Transpose of the dataframe
df_3 = df_1.T
df_3
```

```python
df_4 = df_2.T
df_4
```
