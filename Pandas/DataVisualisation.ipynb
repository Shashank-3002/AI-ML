```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.array(['Q1', 'Q2', 'Q3', 'Q4'])
y = np.array([234.0,254.0,287.0,293.0])

plt.bar(x, y)
plt.xlabel("Quarters")
plt.ylabel("Revenue in USD")
plt.title("Quarterly Revenue")
plt.show()
```

```python
x = np.array(['Q1', 'Q2', 'Q3', 'Q4'])
y = np.array([234.0,254.0,287.0,293.0])

plt.bar(x, y, color = "green", width = 0.4)
plt.xlabel("Quarters")
plt.ylabel("Revenue in USD")
plt.title("Quarterly Revenue")
plt.show()
```

```python
x = np.random.randn(1000)
plt.hist(x, bins = 20,color="orange",density=True)
plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Histogram")
plt.show()
```

```python
x = np.random.randn(10000)
plt.hist(x, bins = 20,color="green",density=True, alpha=0.4)
plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Histogram")
plt.show()
```

```python
x = np.random.randn(10000)*10+3
y = np.random.randn(10000)*10+7
z = np.random.randn(10000)*10+5
plt.boxplot([x,y,z], 0)
plt.title("Box Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
x = np.random.randn(10000)*10+3
y = np.random.randn(10000)*10+7
z = np.random.randn(10000)*10+5
plt.boxplot([x,y,z], 0,vert=False)
plt.title("Box Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)
plt.plot(x, y, color="red")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Sine Wave")
plt.show()
```

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)
plt.plot(x, y, color="red", linestyle="--", marker="o", markersize=5)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Sine Wave")
plt.show()
```

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)
plt.plot(x, y, color="red", linestyle = "-.",markersize=8, markerfacecolor="green", markeredgewidth=2, alpha=0.6)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Sine Wave")
plt.show()
```
