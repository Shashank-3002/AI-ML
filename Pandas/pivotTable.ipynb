```python
import numpy as np
import pandas as pd
```

```python
##Pivot tables
my_dict = {'Size': ['S', 'M', 'L', 'XL', 'XXL', 'S', 'M', 'L', 'XL', 'XXL'],
           'type': ['Polo', 'Polo', 'Polo', 'Polo', 'Polo', 'T-shirt', 'T-shirt', 'T-shirt', 'T-shirt', 'T-shirt'],
            'Price': [500, 600, 700, 800, 900, 300, 400, 500, 600, 700],
            'Quantity': [5, 10, 15, 20, 25, 30, 35, 40, 45, 50]}
my_dict
```

```python
m = pd.DataFrame(my_dict)
m
```

```python
print(m.Price)
print(m.Quantity.mean())
print(m.Price.mean())
print(m.Price.max())
print(m.Quantity.max())
```

```python
pd_1 = m.pivot_table(index='Size', columns='type', values='Price', aggfunc=np.mean)
pd_1
```

```python
pd_2 = m.pivot_table(index='type', columns='Size', values='Price', aggfunc=np.mean)
pd_2
```

```python
pd_3 = m.pivot_table(index='Size', columns='type', values='Quantity', aggfunc=np.mean)
pd_3
```

```python
pd_1.index
```

```python
pd_3.idxmax()
```

```python
pd_1.idxmin()
```

```python
pd_4 = pd.pivot_table(m, index=['Size','type'], columns='Quantity', aggfunc=np.median, fill_value=0)
pd_4
```
