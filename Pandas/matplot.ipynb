```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
%matplotlib inline
```

```python
x = np.random.randn(1000)
y = np.random.randn(1000)
z = np.random.randn(1000)
plt.hist([x, y, z])
plt.show()
```

```python
x = np.random.randn(1000)
y = np.random.randn(1000)
z = np.random.randn(1000)
plt.plot(x,y,z,color='g',marker='o',linestyle='dashed')
plt.show()
```

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)
plt.plot(x, y, color='g', marker='d', linestyle='dashed',alpha=0.5) #alpha is for transparency of line color
plt.title('Sine Wave')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
```

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)
plt.plot(x, y, color='orange', marker='p', linestyle='--',  markeredgecolor='red') #alpha is for transparency of line color
plt.title('Sine Wave')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

```python
fig0 = plt.figure(figsize=(10,5), dpi=100)
ax0 = fig0.add_axes([0,0.5,0.2,0.5])
ax1 = fig0.add_axes([0.3,0.5,0.4,0.4])
ax0.plot(x,y,color='r',marker='x',linestyle='solid', alpha=0.5)
ax1.plot(y,x,color='g',marker='v',linestyle='-.', alpha=0.5)
ax0.set_title('Sine Wave')
ax0.set_xlabel('X-axis')
ax0.set_ylabel('Y-axis')
ax1.set_title('Cosine Wave')
ax1.set_xlabel('Y-axis')
ax1.set_ylabel('X-axis')
plt.show()
```

```python
##Multiple plots in single figure --> used for comparison of different plots
fig1, axes = plt.subplots(nrows=2, ncols=2, figsize=(5,5))
#(0,0)
c = np.random.randn(1000)
axes[0,0].hist(c,color='b',edgecolor='black',bins=120,density=True, alpha=0.5)
axes[0,0].set_title('Histogram')

#(1,0)
axes[1,0].plot(x,y,color='r',marker='^',linestyle='none',markersize = 10, markerfacecolor = 'purple', markeredgecolor = 'black', alpha=0.5)
axes[1,0].set_title('Cosine Wave')
#(0,1) 
a = np.array(['Q1','Q2','Q3','Q4'])
b = np.array([23,45,56,78])
axes[0,1].bar(a,b,color='orange',edgecolor='black', alpha=0.5)
axes[0,1].set_title('Bar Plot')
#(1,1)
axes[1,1].plot(x,y,color='g',marker='o',linestyle='dashed', alpha=0.5)
axes[1,1].set_title('Sine Wave')
plt.tight_layout()
plt.show()
```
