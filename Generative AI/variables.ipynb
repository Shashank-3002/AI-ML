```python
##Python is dynamically typed language, type of a variable is determined at runtime
#Type checking
print(type(age))
print(type(Fname))
print(type(is_student))
```

```python
#type conversion
age_str = str(age) #int to str
print(type(age_str))
print(age_str)
```

```python
age_float = float(age) #int to float
print(type(age_float))
print(age_float)
```

```python
age = '22' #Changing type of age from int to str
print(type(age))
```

```python
age_int = int(age) #str to int
print(type(age_int))
print(age_int)
```

```python
age_float = float(age) #str to float
print(type(age_float))
print(age_float)
```

```python
##Dynamic typing
#Python allows changing the type of a variable by reassigning it to a value of a different type while executing the program.
var = 10 #var is an integer
print(var, type(var))

var = "Hello" #var is now a string
print(var, type(var))

var = 3.14 #var is now a float
print(var, type(var))
```
