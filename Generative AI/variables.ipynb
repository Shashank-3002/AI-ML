```python
##Dynamic typing
#Python allows changing the type of a variable by reassigning it to a value of a different type while executing the program.
var = 10 #var is an integer
print(var, type(var))

var = "Hello" #var is now a string
print(var, type(var))

var = 3.14 #var is now a float
print(var, type(var))
```

```python
##simple calculator
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

#Addition
sum = num1 + num2
print("Sum: ",sum)
#Subtraction
diff = num1 - num2
print("Difference: ",diff)
#Multiplication
prod = num1 * num2
print("Product: ",prod)
#Division
quot = num1 / num2
print("Quotient: ",quot)
#Floor Division
floor_quot = num1 // num2
print("Floor Quotient: ",floor_quot)
#Modulus
mod = num1 % num2
print("Modulus: ",mod)
#Exponentiation
exp = num1 ** num2
print("Exponentiation: ",exp)
#Using math module for square root
sqrt_num1 = np.sqrt(num1)
print("Square root of first number: ",sqrt_num1)
```

```python
#Simple calculator using functions and cases
def add(a, b):
    return a + b
def subtract(a, b):
    return a - b
def multiply(a, b):
    return a * b
def divide(a, b):
    return a / b

print("Select operation:")
print("1.Add")
print("2.Subtract")
print("3.Multiply")
print("4.Divide")
choice = input("Enter choice(1/2/3/4): ")
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
if choice == '1':
    print(num1,"+",num2,"=", add(num1,num2))
elif choice == '2':
    print(num1,"-",num2,"=", subtract(num1,num2))
elif choice == '3':
    print(num1,"*",num2,"=", multiply(num1,num2))
elif choice == '4':
    print(num1,"/",num2,"=", divide(num1,num2))
else:
    print("Invalid input")
```
