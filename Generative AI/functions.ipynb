```python
##Function is a block of code that only runs when it is called. You can pass data, known as parameters, into a function. A function can return data as a result.
def greet(name):
    print("Hello, " + name + ". Good morning!")
greet('Alice')
```

```python
# This is a simple function to check if a number is positive or not
def check_positive(num):
    if num > 0:
        return True
    else:
        return False
check_positive(10)
```

```python
def even_or_odd(num):
    if num % 2 == 0:
        print("Even")
    else:
        print("Odd")
even_or_odd(7)
```

```python
##Function with multiple parameters
def add_numbers(a, b):
    c = a + b
    return c
result = add_numbers(5, 10)
print("The sum is:", result)
```

```python
##Variable length arguments
##Positional arguments --> *args --> takes multiple arguments as a tuple
def sum_all(*args):
    total = 0
    for num in args:
        total += num
    return total
print("Sum:", sum_all(1, 2, 3, 4, 5), "\n", sum_all(10, 20, 30),"\n", sum_all(7, 14, 21, 28, 35, 42))

##Keyword arguments --> **kwargs --> takes multiple arguments as a dictionary
def display_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")
display_info(name="Alice", age=30, city="New York")
display_info(country="USA", profession="Engineer")
```

```python
##Positional and Keyword arguments
def person_info(*args , **kwargs):
    print("Positional arguments:")
    for arg in args:
        print(arg)
    print("\nKeyword arguments:")
    for key, value in kwargs.items():
        print(f"{key}: {value}")

person_info("Alice", 30, city="New York", profession="Engineer")
```

```python
##Return multiple values from a function
def calculate(a, b):
    sum_result = a + b
    product_result = a * b
    return sum_result, product_result
sum_val, prod_val = calculate(5, 10)
print("Sum:", sum_val)
print("Product:", prod_val)
```

```python
##Temperature conversion function
def temperature_conversion(temperature, to_scale):
    if to_scale.lower() == 'celsius':
        return (temperature - 32) * 5.0/9.0 ##F to C
    elif to_scale.lower() == 'fahrenheit':
        return (temperature * 9.0/5.0) + 32 ##C to F
    else:
        return "Invalid scale! Please use 'Celsius' or 'Fahrenheit'."
print("125 Fahrenheit in Celsius:", temperature_conversion(125, 'Celsius'))
print("100 Celsius in Fahrenheit:", temperature_conversion(100, 'Fahrenheit'))
```

```python
##Password strength checker
def password_strength(password):
    if len(password) < 8:
        return "Weak"
    elif any(char.isdigit() for char in password) and any(char.isupper() for char in password) and any(char.islower() for char in password) and any(char in "!@#$%^&*()-_=+[]{}|;:'\",.<>?/`~" for char in password):
        return "Strong"
    else:
        return "Moderate"
print("Password strength for 'Pass123':", password_strength("Pass123"))
print("Password strength for 'Pass123!@#':", password_strength("Pass123!@#"))   
```

```python
##Calculate the total price of items in a shopping cart, including tax and discounts.

def calculate_total(cart):
    total_cost = 0
    for item in cart:
        total_cost += item['price'] * item['quantity']
    return total_cost
cart = [ ##List of dictionaries of items in the cart
    {'name': 'Laptop', 'price': 1000, 'quantity': 1},
    {'name': 'Mouse', 'price': 50, 'quantity': 2},
    {'name': 'Keyboard', 'price': 80, 'quantity': 1}
]
print("Total cost of items in cart:", calculate_total(cart))
```
