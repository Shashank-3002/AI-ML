```python
#Nested conditional statements
num = int(input("Enter a number: "))
if num >= 0:
    if num == 0:
        print("You entered zero")
    elif num > 0:
        print("You entered a positive number")
else:
    print("You entered a negative number")
```

```python
#Nested conditional statements to find even and odd numbers
number = int(input("Enter a number: "))
if number >= 0:
    if number % 2 == 0:
        print("You entered a positive even number")
    else:
        print("You entered a positive odd number")
else:
    if number % 2 == 0:
        print("You entered a negative even number")
    else:
        print("You entered a negative odd number")
```

```python
#Practical example of nested if-else to find whether a year is leap year or not
year = int(input("Enter a year: "))
if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print(f"{year} is a leap year")
        else:
            print(f"{year} is not a leap year")
    else:
        print(f"{year} is a leap year")
else:
    print(f"{year} is not a leap year")
```

```python
#Simple calculator using nested if-else
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
operation = input("Enter operation (+, -, *, /): ")
if operation == '+':
    print(f"{num1} + {num2} = {num1 + num2}")
elif operation == '-':
    print(f"{num1} - {num2} = {num1 - num2}")
elif operation == '*':
    print(f"{num1} * {num2} = {num1 * num2}")
elif operation == '/':
    if num2 != 0:
        print(f"{num1} / {num2} = {num1 / num2}")
    else:
        print("Error: Division by zero")
else:
    print("Invalid operation")
```

```python
###Loop control statements in python are used to change the execution flow of loops. There are three main loop control statements in python: break, continue, and pass.
#Break statement is used to exit a loop prematurely when a certain condition is met
for i in range(10):
    if i == 5:
        break
    print(i)
```

```python
# --- IGNORE ---
#continue statement is used to skip the current iteration of a loop and move to the next iteration
for i in range(10):
    if i % 2 == 0: #If the number is even then skip it and continue to the next iteration
        continue
    print(i) #This will print only odd numbers from 0 to 9
```

```python
#pass statement is a null operation; it is used when a statement is syntactically required but no action is needed
# --- IGNORE ---
for i in range(5):
    if i == 3:
        print("This is pass statement example")
        pass
    print(i)
```

```python
#Nested loops in python: Example of Bubble sort algorithm
# --- IGNORE ---
arr = [64, 34, 25, 12, 22, 11, 90]
n = len(arr)
for i in range(n):
    for j in range(0, n-i-1):
        if arr[j] > arr[j+1]:
            temp = arr[j]
            arr[j] = arr[j+1]
            arr[j+1] = temp
print("Sorted array is:", arr)
```

```python
#Nested loops in python
for i in range(3):
    print(f"Start of outer loop i = {i}")
    for j in range(3):
        print(f"i = {i}, j = {j}")
    print(f"End of outer loop i = {i}\n")
```

```python
#Example of sum of first n natural numbers using while loop
n = int(input("Enter a positive integer: "))
sum = 0
count = 1
while count <= n:
    sum += count #sum = sum + count
    count += 1 #count = count + 1
print(f"The sum of first {n} natural numbers is {sum}")
```

```python
#Example of sum of first n natural numbers using for loop
n = int(input("Enter a number: "))
sum = 0
for i in range(1, n+1):
    sum += i #sum = sum + i
print(f"The sum of first {n} natural numbers is {sum}")
```

```python
#Example of Prime number using for loop
num = int(input("Enter a number: "))
for i in range(2, num // 2 + 1):
    if (num % i == 0):
        print(f"{num} is not a prime number")
        break
else:
    print(f"{num} is a prime number")
```

```python
#Print prime numbers in a given range
import numpy as np

nums = int(input("Enter a number: "))
prime_list = []

for num in range(2, nums):
    for i in range(2, num // 2 + 1):
        if (num % i == 0):
            break
    else:
        prime_list.append(num)

print("Prime numbers:", prime_list)
print("Total number of prime numbers:", len(prime_list))
```
