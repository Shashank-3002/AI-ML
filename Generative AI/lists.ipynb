```python
#Lists
lst = [10,20,30,40,50]
print(lst)
print(type(lst))
```

```python
names_list = ["Alice", "Bob", "Charlie", "David", 10,20,50,60]
print(names_list)
print(type(names_list))#Prints the type of the list
print(type(names_list[3])) #Prints the data type of the element at index 3
print(type(names_list[6])) #Prints the data type of the element at index 6
```

```python
mixed_list = [1, "Shashank", 'Python', 3.142, True]
print(mixed_list)
print(type(mixed_list))
# type(mixed_list) --> Print the output as list only
print(type(mixed_list[4]))
```
