Data Types in Python

Basic Data Types are:
1. Integers
2. Floating point nummbers
3. Strings
4. Booleans

```python
#1. Integer
age = 22
print(type(age))
print(age)
#2. Float
height = 5.9
print(type(height))
print(height)
#3. String
name = "John"
print(type(name))
print(name)
#4. Boolean
is_student = True
print(type(is_student))
print(is_student)
```

Advanced Data Types are:
1. Lists --> represented using [ ]
2. Tuples --> represented using paranthesis ( )
3. Sets --> represented using flower brackets { }
4. Dictionary --> It is represented like a set but every literal(value) is assigned to a 'key'

```python
#1. List
fruits = ["apple", "banana", "cherry"]
print(type(fruits))
print(fruits)
#2. Tuples
names = ("Shashank","Lohit","Sai Prem","Prateek","Pramod")
print(type(names))
print(names)
#3. Sets
vegetables = {"Onion", "Cucumber", "Carrot"}
print(type(vegetables))
print(vegetables)
vowels = {'a','e','i','o','u'}
print(type(vowels))
print(vowels)
#4. Dictionary --> A dictionary is a set with a key and has a value assigned to each key
alphabets = {'a':"apple", 'b': "bat", 'c': "Cat"}
print(type(alphabets))
print(alphabets)
```
