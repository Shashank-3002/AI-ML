```python
#Matrix Operation on the Arrays
import numpy as np
```

```python
A = np.array([[1,2,4,5]])
A
```

```python
#Transpose of the given matrix array
B= A.transpose() #or else we can also use np.transpose(A)
B
```

```python
#Dot product of two arrays
A.dot(B)
```

```python
#Another method for the same multiplication
np.dot(A,B)
```
