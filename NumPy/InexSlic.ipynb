```python
import numpy as np
```

```python
nums = np.array([5,10,15,20,25,30,35,40,45,50])
```

```python
nums
```

```python
nums[1]
```

```python
print(nums[4])
```

```python
#Generating a 3x3 Matrix and extracting elements using indexing
nums10 = np.arange(10,100,10).reshape(3,3)
```

```python
print(nums10[1])
```

```python
print(nums10[1,2])
```

```python
#Insering the values index in a matrix
nums10[2,2] = 100
```

```python
nums10
```

```python
nums10[2]
```

```python
#Accesing multiple elements of diff arrays in the given 3D Array
nums10[[0,2],[2,1]]
```

```python
#Accesing using the condition
nums10[nums10>40]
```

```python
nums10[nums10%40 == 0]
```
