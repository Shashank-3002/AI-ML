```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder, OneHotEncoder
import pickle

X = df.drop('addicted_label', axis=1)  # replace with actual column
y = df['addicted_label']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
import pandas as pd

# Scaling the features
scaler = StandardScaler()

# Identify identifier columns to drop
identifier_cols = ['transaction_id', 'user_id']

# Identify categorical columns for OneHotEncoding
categorical_cols = X_train.select_dtypes(exclude=['number']).columns.tolist()
categorical_cols = [col for col in categorical_cols if col not in identifier_cols]

# Drop identifier columns from X_train and X_test
X_train_processed = X_train.drop(columns=identifier_cols)
X_test_processed = X_test.drop(columns=identifier_cols)

# Apply OneHotEncoder to categorical columns
ohe = OneHotEncoder(handle_unknown='ignore', sparse_output=False)

# Fit and transform on X_train for categorical columns
encoded_features_train = ohe.fit_transform(X_train_processed[categorical_cols])
encoded_df_train = pd.DataFrame(encoded_features_train, columns=ohe.get_feature_names_out(categorical_cols), index=X_train_processed.index)

# Transform X_test for categorical columns
encoded_features_test = ohe.transform(X_test_processed[categorical_cols])
encoded_df_test = pd.DataFrame(encoded_features_test, columns=ohe.get_feature_names_out(categorical_cols), index=X_test_processed.index)

# Drop original categorical columns and concatenate encoded features
X_train_processed = X_train_processed.drop(columns=categorical_cols)
X_train_processed = pd.concat([X_train_processed, encoded_df_train], axis=1)

X_test_processed = X_test_processed.drop(columns=categorical_cols)
X_test_processed = pd.concat([X_test_processed, encoded_df_test], axis=1)

# Now apply StandardScaler to the processed numerical dataframes
X_train = scaler.fit_transform(X_train_processed)
X_test = scaler.transform(X_test_processed)
```
