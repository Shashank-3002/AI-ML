```python
from scipy.stats import bernoulli
```

```python
# Probability of success
p = 0.3
#probability Mass Function
print("Probabilty of success P(X=1): ",bernoulli.pmf(1, p))  # Probability of success
print("Probabilty of failure P(X=0): ",bernoulli.pmf(0, p))  # Probability of failure
#Random sample generation
samples= bernoulli.rvs(p, size=10)
print("Random Samples: ", samples)
```

```python
#Visualization of Bernoulli PMF
import matplotlib.pyplot as plt
import numpy as np

p = 0.7
x = [0, 1]
pmf = bernoulli.pmf(x, p)

plt.stem(x, pmf, basefmt=" ", linefmt='b-', markerfmt='bo')
plt.xticks(x)
plt.xlabel('Outcomes')
plt.ylabel('Probability P(X=x)')
plt.title('Bernoulli Distribution PMF (p={})'.format(p))
plt.ylim(0, 1)
plt.grid()
```
