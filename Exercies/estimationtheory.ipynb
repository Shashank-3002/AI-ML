```python
import numpy as np
from scipy.stats import norm,t,chi2
```

### Point Estimation

```python
def point_estimate(data):
    xbar = np.mean(data) # Sample Mean
    s = np.std(data, ddof=1) # Sample Standard Deviation
    return xbar, s
```

### Confidence Interval — MEAN (σ KNOWN)

```python
def ci_mean_sigma_known(xbar, sigma, n, confidence=0.95):
   alpha = 1 - confidence
   z = norm.ppf(1 - alpha/2)
   me = z * (sigma / np.sqrt(n))
   return xbar - me, xbar + me
```

### CONFIDENCE INTERVAL — MEAN (σ UNKNOWN)

```python
def ci_mean_sigma_unknown(xbar, s, n, confidence=0.95):
    n = len(data)
    s = np.std(data, ddof=1)
    alpha = 1 - confidence
    t_crit = t.ppf(1 - alpha/2, df=n-1)
    me = t_crit * (s / np.sqrt(n))
    return xbar - me, xbar + me
```

### Confidence Interval - Proportion

```python
def ci_proportion(x,n,confidence=0.95):
    p_hat = x / n
    alpha = 1 - confidence
    z = norm.ppf(1 - alpha/2)
    me = z * np.sqrt((p_hat * (1 - p_hat)) / n)
    return p_hat - me, p_hat + me
```

### CONFIDENCE INTERVAL — VARIANCE

```python
def ci_variance(data, confidence=0.95):
    n = len(data)
    s2 = np.var(data, ddof=1)
    alpha = 1 - confidence
    chi2_lower = chi2.ppf(alpha/2, df=n-1)
    chi2_upper = chi2.ppf(1 - alpha/2, df=n-1)
    lower_bound = (n - 1) * s2 / chi2_upper
    upper_bound = (n - 1) * s2 / chi2_lower
    return lower_bound, upper_bound
```

### PROPERTIES OF ESTIMATORS — BIAS, VARIANCE, MSE

```python
def estimator_properties(estimator, data, true_value):
    estimates = [estimator(np.random.choice(data, size=len(data), replace=True)) for _ in range(1000)]
    estimates = np.array(estimates)
    bias = np.mean(estimates) - true_value
    variance = np.var(estimates)
    mse = bias**2 + variance
    return bias, variance, mse
```

### Simulation - Compare two estimators of variance

```python
def simulate_estimators(sample_size=30, iterations=1000):
    biased_estimates = []
    unbiased_estimates = []
    for _ in range(iterations):
        sample = np.random.normal(loc=50, scale=10, size=sample_size)
        # Biased estimator (dividing by n)
        biased_est = np.mean((sample - np.mean(sample))**2)
        # Unbiased estimator (dividing by n-1)
        unbiased_est = np.var(sample, ddof=1)
        biased_estimates.append(biased_est)
        unbiased_estimates.append(unbiased_est)

        true_sigma2 = 10**2  # True variance (10^2)
        biased_properties = estimator_properties(biased_estimates, true_sigma2)
        unbiased_properties = estimator_properties(unbiased_estimates, true_sigma2)

        return biased_properties, unbiased_properties
```
