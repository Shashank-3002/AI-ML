```python
import numpy as np

a = np.array([5, 3, 4, 4])
b = np.array([3, 1, 2, 3])
c = np.array([4, 3, 4, 3])

def euclidean(u, v):
    return np.linalg.norm(u - v)

print("a-b:", euclidean(a, b))
print("a-c:", euclidean(a, c))
print("b-c:", euclidean(b, c))
```

```python
from numpy import dot
from numpy.linalg import norm

def cosine(u, v):
    return 1 - dot(u, v) / (norm(u) * norm(v))

print("a-b:", cosine(a, b))
print("a-c:", cosine(a, c))
print("b-c:", cosine(b, c))
```

```python
import sympy as sp

x, y, k, a, b = sp.symbols('x y k a b')

# (a)
f1 = x**3 - 1
df1 = sp.diff(f1, x)
print("f'(x) =", df1)
```

```python
# (b)
f2 = sp.log(x**2 - 3*k)
df2 = sp.diff(f2, x)
print("f'(x) =", df2)
```

```python
f3 = sp.exp(a*x**b)
df3 = sp.diff(f3, x)
print("f'(x) =", df3)
```

```python
# (d) Partial derivatives
f4 = sp.exp(x**2 + 2*y**2)
fx = sp.diff(f4, x)
fy = sp.diff(f4, y)
fxx = sp.diff(fx, x)
fxy = sp.diff(fx, y)
fyx = sp.diff(fy, x)
fyy = sp.diff(fy, y)
print("f_x =", fx)
print("f_y =", fy)
print("f_xx =", fxx)
print("f_xy =", fxy)
print("f_yx =", fyx)
print("f_yy =", fyy)
```

### Boston House Price Problem: Linear Prediction Model (Ax = b) with 4 Features and 4 Data Points


```python
import pandas as pd
import numpy as np
boston = pd.read_csv('C:\\Users\\shash\\OneDrive\\Desktop\\SamsungInnovationCampus-AI-ML\\SamsungInnovationCampus-AI-ML\\DataSets\\data_boston.csv')
A = boston.iloc[:4][['CRIM', 'NOX', 'RM', 'AGE']].values
b = boston.iloc[:4]['PRICE'].values
x = np.linalg.lstsq(A, b, rcond=None)[0]
print('Weight vector x:', x)
```

#### Interpretation of the weight vector x for CRIM, NOX, RM, AGE
- Positive weights: factors that increase house price.
- Negative weights: factors that decrease house price.
- Example: If CRIM weight is negative, house price is in inverse proportion to crime rate (CRIM).

### Boston House Price Problem: Least Squares Solution with All Features and Data


```python
X = boston.drop('PRICE', axis=1).values
y = boston['PRICE'].values
w = np.linalg.lstsq(X, y, rcond=None)[0]
print('Weight vector w:', w)
```

#### Interpretation of the weight vector w for all features
- Positive weights: factors that increase house price.
- Negative weights: factors that decrease house price.
- Example: If CRIM weight is negative, house price is in inverse proportion to crime rate (CRIM).


#### Difference between the two results
The full least-squares solution uses all data and features, so weights are more accurate and reflect all factors, unlike the simplified 4-row/4-feature version.
