```python
import sympy as sp
# (a) f(x) = (x**2+2)**3
x = sp.Symbol('x')
fx = (x**2+2)**3
sp.diff(fx,x)
```

```python
x = sp.Symbol('x')
gx = 1/(x**4+1)**2
sp.diff(gx,x)
```

```python
x  = sp.Symbol('x')
gx  = 3*sp.log(x,2) - x*sp.log(x)
print("g'(x) = ", sp.simplify(sp.diff(gx,x)))
```
