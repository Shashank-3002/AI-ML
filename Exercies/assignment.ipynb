```python
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
%matplotlib inline

df = pd.read_csv('C:\\Users\\shash\\OneDrive\\Desktop\\SamsungInnovationCampus-AI-ML\\SamsungInnovationCampus-AI-ML\\DataSets\\data_coffeeshop.csv', header='infer',na_values=[' '])
df.head(5)
```

```python
df.shape
```

```python
# 1) Frequency table and line plot for 'yearOfStart'
import matplotlib.pyplot as plt
import pandas as pd

# Frequency table
freq_table = df['Year of Start'].value_counts().sort_index()
print("Frequency Table for 'yearOfStart':\n", freq_table)

# Line plot from 1997 to 2014
plt.figure(figsize=(10,5))
freq_table.loc[1997:2014].plot(kind='line', marker='o')
plt.title("Frequency of Businesses by Year of Start (1997-2014)")
plt.xlabel("Year of Start")
plt.ylabel("Frequency")
plt.xlim(1997, 2014)
plt.grid(True)
plt.show()
```

```python
# 2) Split by 'currentState' and plot yearly trends
plt.figure(figsize=(10,5))
for state in ['In', 'Out']:
    state_freq = df[df['Current State'] == state]['Year of Start'].value_counts().sort_index()
    state_freq = state_freq.loc[1997:2014]  # restrict to 1997-2014
    plt.plot(state_freq.index, state_freq.values, marker='o', label=state)
plt.title("Yearly Trend of Business Start by State (1997-2014)")
plt.xlabel("Year of Start")
plt.ylabel("Frequency")
plt.xlim(1997, 2014)
plt.legend(title='Business State')
plt.grid(True)
plt.show()
```

### 3) Trend Description
From the above plots:
- The frequency of business starts generally fluctuates between 1997 and 2014, with some years showing peaks and others dips.
- When split by 'currentState', both 'In' and 'Out' of business follow similar trends, but the number of businesses that are 'Out' may increase in certain years, indicating possible business closures after initial peaks.
- Overall, the trend may suggest periods of increased entrepreneurship followed by stabilization or decline, possibly due to economic or market factors.
