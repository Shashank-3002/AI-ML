```python
import pandas as pd
import glob

outputs = []

# Correct file-matching pattern
files = sorted(glob.glob("Fig.4(*).out"))

created_files = []

for fpath in files:
    rows = []
    
    with open(fpath) as f:
        for line in f:
            parts = line.split()

            # Expecting ETA, F, G → 3 numeric columns
            if len(parts) >= 3 and parts[0] != "ETA":
                try:
                    eta = float(parts[0])
                    F   = float(parts[1])
                    G   = float(parts[2])
                    rows.append([eta, F, G])
                except ValueError:
                    # skip non-numeric header/lines
                    pass

    # Create DataFrame
    df = pd.DataFrame(rows, columns=['ETA', 'F', 'G'])

    # Save to CSV
    csv_path = fpath.replace('.out', '.csv')
    df.to_csv(csv_path, index=False)

    created_files.append(csv_path)

created_files
```
