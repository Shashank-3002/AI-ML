```python
import pandas as pd
import numpy as np
from scipy.optimize import least_squares

# Load dataset
df = pd.read_csv("C:\\Users\\admin\\Desktop\\Puneeth\\Research\\Dr P M Patil Sir\\Tetra Cone\\Gradients\\Fig 4 & 5 Epsilon and n\\Grad-4-pinn-results\\Cf_long_raw.csv")

x = df['Eta'].values
y = df['Cf_val'].values

# Nonlinear model
def model(p, x):
    a, b, c = p
    return a * np.exp(-b * x) + c

# Residuals
def residuals(p, x, y):
    return model(p, x) - y

# Initial guess
p0 = [1, 1, 0]

# Levenberg–Marquardt optimization
result = least_squares(residuals, p0, args=(x, y), method='lm')

print("Optimized Parameters:", result.x)
```

```python
#Plot the graph
import matplotlib.pyplot as plt
plt.scatter(x, y, label='Data', color='red', s=10)
plt.plot(x, model(result.x, x), label='Fitted Model', color='blue')
plt.xlabel('Eta')
plt.ylabel('Cf_val')
plt.legend()
plt.show()
```
