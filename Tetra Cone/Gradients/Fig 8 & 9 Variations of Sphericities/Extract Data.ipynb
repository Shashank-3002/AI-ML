```python
import pandas as pd
import glob

outputs = []

# File-matching pattern
files = sorted(glob.glob("Grad.6(*).out"))

created_files = []

for fpath in files:
    rows = []
    
    with open(fpath) as f:
        for line in f:
            parts = line.split()

            # Expecting 2 numeric values: ETA, Cf
            if len(parts) >= 2:
                try:
                    eta = float(parts[0])
                    pi   = float(parts[1])
                    #G   = float(parts[2])
                    rows.append([eta, pi])
                except ValueError:
                    # skip non-numeric lines
                    pass

    # Create DataFrame with NEW headers
    df = pd.DataFrame(rows, columns=['ETA', 'Pi'])

    # Save CSV
    csv_path = fpath.replace('.out', '.csv')
    df.to_csv(csv_path, index=False)

    created_files.append(csv_path)

created_files
```
